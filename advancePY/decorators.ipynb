```python
def decorator(func):#function as argument
    #inner function 
    def wrapper():
        print("before function call")
        func() #actual function call
        print("after function call")
    return wrapper #function as value
```

```python
@decorator
def greet():
    print("namaste")
    
```

```python
greet()
```

```python
def name(func):
    def wrapper():
        print("before")
        print(func().upper())
        print("after")
    return wrapper
```

```python
@name
def nam():
    print("enter your name")
    name=input()
    return name
    
    
```

```python

nam()
```
