# ITERATORS 

```python
class Fibonacci:
    def __init__(self,limit=10):
        self.limit=limit
        self.current=0
        self.next=1
        self.step=0
    def __iter__(self):
        return self
    def __next__(self):
        if self.step>=self.limit:
            raise StopIteration 
        result=self.current
        self.current,self.next=self.next,self.current+self.next
        self.step+=1
        return result 
        
```

```python
f=Fibonacci(11)
```

```python
print(f)
```

```python
for num in f:
    print(num)
```
