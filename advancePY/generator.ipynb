# GENERATOR 


```python
def number_gen(start,end):
    for num in range(start,end+1):
        yield num
```

```python
lala=number_gen(1,10)
```

```python
next(lala)
next(lala)
next(lala)
```

```python
for i in lala:
    print(i)
```

```python
def check(num):
    if num<=0:
        return False 
    for i in range(2,int(num**0.5)+1):
        return False
    return True
    
```

```python
def prime(limit=10):
    n,i=1,1
    while i<=limit:
        if check(n):
            i+=1
            yield n
        n+=1
```

```python
for i in prime(100):
    print(i)
```
