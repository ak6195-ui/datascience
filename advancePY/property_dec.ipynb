```python
class Person():
    def __init__(self,name):
        self.__fullname=name
    @property
    def name(self):
        return  self.__fullname
    @name.setter
    def name(self,value):
        print("setting name")
        self.__fullname=value
        

```

```python
person=Person("abi")
```

```python
person.name
```

```python
person.name="khadka"
```

```python
person.name
```

# DELETER

```python
class Person():
    def __init__(self,name):
        self.__fullname=name
    @property
    def name(self):
        return  self.__fullname
    @name.setter
    def name(self,value):
        print("setting name")
        self.__fullname=value
    @name.deleter
    def  name(self):
        print("deleting")
        del self.__fullname

```

```python
person=Person("abi kahdka")
```

```python
person.name
```

```python
person.name="ola baby"
```

```python
person.name
```

```python
del person.name
```

```python
person.name="abi khadaka"
```

```python
import time
import random 
def timeit(func):
    def wrapper():
        result=func()
        return result
    return wrapper
```

```python
@timeit
def tim():
    end=time.time()+random.uniform(1,10)
    result=0
    while  time.time()<end:
        result+=random.randint(1,100)
    return result
```
