```python
import seaborn as sns 
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
tips=sns.load_dataset("tips")
```

```python
tips
```

```python
# SCATTER PLOT ->axes  level function
sns.scatterplot(data=tips,x="total_bill",y="tip",hue="sex",style="time")
```

```python
# RELPLOT IS FIGURE LEVEL FUNCTION 
sns.relplot(data=tips,x="total_bill",y="tip",kind="scatter",hue="sex")
```

```python
tips 
```

```python
sns.relplot(data=tips,x="total_bill",y="tip",kind="scatter",hue="sex",style="smoker",size="size")
```

```python
# LINE PLOT->AXES LEVEL 
gap=px.data.gapminder()
gap
```

```python
gap.groupby(["country","continent"]).agg(total_life_exx=("lifeExp","sum"),average_population=("pop","mean"))
```

```python
nep=gap[gap["country"]=="Nepal"]
sns.lineplot(data=nep,x="year",y="lifeExp")
```

```python
# using figure level function 
sns.relplot(data=nep,x="year",y="lifeExp",kind="line")
```

```python
count=gap[gap["country"].isin(["Germany","Nepal","Brazil"])]
```

```python
sns.relplot(kind="line",data=count,x="year",y="lifeExp",hue="country",style="continent")
```

```python
# FACET PLOT 
sns.relplot(data=tips,x="total_bill",y="tip",kind="scatter",col="sex",hue="smoker",style="time",row="day")
```

```python
#col wrap 
sns.relplot(data=gap,x="lifeExp",y="gdpPercap",kind="scatter",col="year",col_wrap=2)
```

```python
titanic=sns.load_dataset("titanic")
titanic
sns.displot(data=titanic,x="age",hue="sex",element="step")
```

```python
sns.displot(data=titanic,x="age",col="sex",row="class")
```

```python
# KDE PLOT KERNEL DENSITY ESTIMATION 
sns.kdeplot(data=tips,x="total_bill")
```

```python
sns.displot(data=titanic,x="age",kind="kde",hue="sex",fill=True)
```

```python
# RUG PLOT SHOWS SAME DISTRIBUTION IN ANOTHER WAY 
sns.kdeplot(data=titanic,x="age",hue="sex",fill=True)
sns.rugplot(data=titanic,x="age")
```
