# MACHINE LEARNING 
MACHINE LEARNING IS A BRANCH OF AI THAT INVOLVES DEVELOPING ALGORITHMS AND STATISTICAL MODELS ENABLECOMPUTERS TO AUTOMATICALLY LEARN FROM DATA WITHOUT BEING EXPLICITLY PROGRAMMED.IN ML ALGORITHMS USE DATA TO IDENTIFY PATTERNS AND MAKE DESICIONS WITH MINIMUM HUMAN INTERVENTION .MACHINE LEARNING IS USED IN A WIDE RANGE OF APPLICATIONS FROM SELF DRIVING CARS AND FACIAL RECOGNITION  TECHNOLOGY TO MEDICAL DIGNOSIS AND PROCESSING.

# MAJOR APPROACHES OF ML 
1)SUPERVISED LEARNING:
THIS APPROACH INVOLVES USING LABELED DATASET TO TRAIN A MODEL THAT CAN PREDICT OUTPUT VALUES FOR NEW INPUT DATA.IT IS CALLED SUPERVISED BECAUSE THE TRAINNING DATA INCLUDES THE CORRECT ANSWERS AND THE GOAL IS FOR MODEL TO LEARN HOW TO PRODUCE THE CORRECT OUTPUT FOR NEW UNSEEN DATA


2)UNSUOERVISED LEARNING:
THIS APPROACH INVOLVED USING UNLABELED DATASET TO TRAIN A MODEL THAT CAN IDENTIFY PATTERNS OR STRUCTURE IN THE DATA.THERE IS NO CORRECT ANSWER TO LEARN FROM ,SO THE MODEL IS TYPICALLY USED FOR TASK LIKE SLUSTERING ,WHERE SIMILAR DATA POINTS ARE GROUPED TOGETHER 

3)REINFORCEMENT LEARNING :
THIS APPROACH INVOLVES TRANNIG A MODEL TO MAKE DECISIONS BASED ON REWARD OR PENALTIES RECEIVED IN RESPONSE TO ITS ACTIONS.THE GOAL IS FOR THE MODEL TO LEARN TO TAKE ACTIONS THAT MAXIMIZE ITS CUMULATIVE REWARD OVER TIME .THIS APPROACH IS OFTEN USED IN AREAS LIKE GAME-PLAYING AND ROBOTICS WHERE AN AGENT MUST LEARN TO NAVIGATE AN ENVIRONMENT AND MAKE DECISIONS BASED ON FEEDBACK.
