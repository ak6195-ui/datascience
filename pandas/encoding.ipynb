```python
from sklearn.preprocessing import OneHotEncoder
import numpy as np
# so  OneHotEncoder is an class so to use its property we have to create its object 
```

```python
one_hot_encoder=OneHotEncoder(sparse_output=False)
# we have created the object name one_hot_encoder and aso in parameter we have done sparse_output=False beacuse to take output in arrays and if we had done sparse_output=True then we would get output in matrix
```

```python
encoder_gender=one_hot_encoder.fit_transform(titanic[["Sex"]])
#also WE HAVE USED one_hot_encoder.fit_transform(titanic[["Sex"]]) -> because onehotencoder dont accept the pandas series like 1-D objects so we made it 2D by doimg [[]]
```

```python
encoder_gender
```

```python
one_hot_encoder.categories_
# so,one_hot_encoder.categories_ gives the idea about the number used and what that number represnts 
```

```python
one_hot_encoder.inverse_transform(encoder_gender)
# so, onehotencoder.inverse_transform(encoder_gender),does gives the meaning behind that numbers ir what actually that numbers stores
# also, we can write the output also like one_hot_encoder.inverse_transform([1,0],[0,1]) like this and the output will be given according to what number that actually represents
```

```python
df=pd.DataFrame(encoder_gender,columns=one_hot_encoder.categories_)
```

```python
titanic=pd.concat([titanic,df],axis=1)
titanic
```
