```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
import scipy.stats as stats
```

# Z SCORE 
it shows that how many standard deviation is the data far from the mean value 

```python
x= np.linspace(-3,3,20000)
plt.figure(figsize=(8,6))
plt.plot(x,stats.norm.pdf(x),"black",lw=2,label="STANDARD NORMAL DISTRIBUTION")
plt.title("STANDARD NPORMAL DISTRIBUTION")
plt.xlabel("Z-SCORE")
plt.ylabel("DENSITY")
plt.legend()
plt.show()
```

```python
# calculating and interpreting z-score
sample=[150,155,160,165,170,175,180]
sample_mean=np.mean(sample)
sample_std=np.std(sample,ddof=1)
z_score=(sample-sample_mean)/(sample_std)
for i,z in enumerate(z_score):
    print(f"data point {i+1}. {sample[i]} has z-score={z:.3f}")
```
