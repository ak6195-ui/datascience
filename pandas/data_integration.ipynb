```python
import pandas as pd 
```

# CONCATINATE 

```python
df1=pd.DataFrame({"acc":[1,2,3],"name":["a","b","c"],"age":[2,3,4]})
df2=pd.DataFrame({"principle":[2,3,5],"interest":[3,4,5]})
```

```python
pd.concat([df1,df2],axis=1)
```

```python
pd.concat([df1,df2],axis=0,ignore_index=True)
```

# MERGE 

```python
df1=pd.DataFrame({"key":["A","B","C","D"],"inv_val":[99,88,77,66],"acc":[1,2,3,4]})

df2=pd.DataFrame({"key":["E","B","C","Z"],"hel_ratio":[99,88,77,66],"acc":[11,22,33,44]})
```

```python
df1,df2
```

```python
pd.merge(left=df1,right=df2,how="left",on="key")
```

```python
pd.merge(left=df1,right=df2,how="right",on="key",suffixes=("acc_l","acc_r"))
```

```python
df1,df2
```

```python
pd.merge(left=df1,right=df2,how="outer",on="key")
```

```python

df3=pd.DataFrame({"key":["E","F","G","Z"],"hel_ratio":[99,88,77,66],"acc":[11,22,33,44]})
```

```python

df4=pd.DataFrame({"key":["E","I","G","K"],"hel_ratio":[99,88,77,66],"acc":[11,22,33,44]})
```

```python
df3,df4
```

```python
pd.merge(left=df3,right=df4,on="key",how="right",suffixes=("df3","df4"))
pd.merge(left=df3,right=df4,on="key",how="outer",suffixes=("df3","df4"),indicator=True)
```

```python
df3.merge(df4,on="key")
```

# JOIN 

```python
df3,df4
```

```python
df3.join(df2,how="inner",lsuffix="left",rsuffix="right")
```

```python
df3,df4
```

```python
df4.join(df3,how="inner",lsuffix="_left",rsuffix="_right")
```

```python
df1.join(df2,how="inner",lsuffix="_left",rsuffix="_right")
```

# SAVE DATA TO CSV

```python
titanic=pd.read_csv("train.csv",nrows=10)
titanic.Age.dropna(inplace=True)
titanic.isnull().sum()
titanic["Age"]=titanic.Age.fillna(titanic.groupby("Sex")["Age"].transform("mean"))
```

```python
titanic["Cabin"]=titanic.Cabin.fillna(titanic.Cabin.mode(),inplace=True)
```

```python
titanic["Cabin"]=titanic["Cabin"].ffill()
```

```python
titanic.to_csv("cleaned_titanic.csv")
```
