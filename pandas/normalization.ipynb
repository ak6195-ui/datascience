# MIN MAX NORMILIZATION 
THIS METHOD SCALES DATA TO FIXED RAGE,TYPICALLY BETWEEN 0 AND 1.


x_norm=(x-x_min)/(x_max-xmin)

```python
final_df=pd.read_csv("final_data.csv")
min_fare=final_df.Fare.min()
max_fare=final_df.Fare.max()
```

```python
norm_fare=(final_df.Fare-min_fare)/(max_fare-min_fare)
```

# LOG TRANSFORMATION
for value starting from low and sudden rise in the value for example 0 to 1 million 

```python
import numpy as np
peak=5*10e2
n=250
k=np.log(peak)
growth=np.exp(np.linspace(1,k,n))
decay=np.exp(np.linspace(k,1,n))
data=np.concatenate([growth,decay[1:]])
```

```python
pd.DataFrame(data,columns=["Data"]).plot(kind="line")
```

```python
pd.DataFrame(np.log(data),columns=["Data"]).plot(kind="line")
```

# POWER TRANSFORMATION 
to adjust the skewness of data to more symmetric and normally distributed

NORM=SIGN(X)*ABS(X)**A

```python
from scipy.stats import skewnorm
data=pd.DataFrame(skewnorm.rvs(25,size=200))
data.plot.kde()
```

```python
pow_trans=lambda series,pow_fact : np.sign(series) * np.power(np.abs(series),pow_fact)
```

```python
trans_data= pow_trans(data,0.25).plot.kde()
```
