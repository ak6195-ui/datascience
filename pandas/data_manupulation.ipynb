```python
titanic.Age.plot(kind="box")
```

```python
q1=titanic.Age.quantile(q=0.25)
q3=titanic.Age.quantile(q=0.75)
IQR=q3-q1
```

```python
maximum= q3+(1.5*IQR)
minimum=q1-(1.5*IQR)
```

```python
maximum,minimum
```

```python
new_titanic2=pd.read_csv("train.csv")
new_titanic2[(new_titanic2["Age"]<maximum)|(new_titanic2["Age"]>minimum)]
```

```python
# import scipy
# scipy.stats.zscore(titanic["Age"])
```

```python
titanic["z_score"]=titanic["Age"].apply(lambda x:f"{(x-titanic["Age"].mean())/(titanic["Age"].std()):.4f}")
```

```python
titanic.dropna(inplace=True)
```

```python
new_titanic2["z_score"]=new_titanic2["Age"].apply(lambda x: (x-new_titanic2["Age"].mean())/new_titanic2["Age"].std())
```

```python
new_titanic2["z_score"].dropna(inplace=True)
```

```python
new_titanic2[new_titanic2["z_score"]>2.5]
```

```python
new_titanic["z_score_fare"]=new_titanic["Fare"].apply(lambda x:(x-new_titanic["Fare"].mean())/(new_titanic["Fare"].std()))
```
